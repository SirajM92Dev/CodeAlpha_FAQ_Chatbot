# FAQ Chatbot — Online Payment App

A simple rule-based FAQ chatbot that uses **TF-IDF** vectorization and **cosine similarity** to match a user's question to the closest FAQ and return its answer.

**Features**
- Interactive chat loop (type `exit` to quit)
- Greeting and farewell messages
- Handles empty input
- 18 sample FAQs covering common online store queries

```python
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
import string

# Download required NLTK data (only downloads if not already present)
# nltk.download('punkt', quiet=True)
# nltk.download('punkt_tab', quiet=True)
# nltk.download('stopwords', quiet=True)
```

```python
# FAQs and Answers (Topic: Online payment app)
FAQs = [
    "What payment methods do you accept?",
    "How can I track my order?",
    "What is your return policy?",
    "How do I request a refund?",
    "How long does delivery take?",
    "Can I cancel my order?",
    "How do I reset my account password?",
    "Do you offer international shipping?",
    "How can I contact customer support?",
    "Is cash on delivery available?",
    "How do I create an account?",
    "Is it safe to save my card details?",
    "Can I change my delivery address after ordering?",
    "Do you charge extra for online payments?",
    "What should I do if my payment fails?",
    "How do I apply a discount coupon?",
    "Can I pay using a mobile wallet?",
    "How do I delete my account?"
]

Answers = [
    "We accept credit/debit cards, bank transfers, and cash on delivery.",
    "You can track your order from the 'My Orders' section using your order ID.",
    "Products can be returned within 30 days if they are unused and in their original packaging.",
    "Go to your order history, select the order, and click 'Request Refund'.",
    "Standard delivery takes 3–7 business days, depending on your location.",
    "Yes, you can cancel your order before it has been shipped.",
    "Click 'Forgot Password' on the login page and follow the instructions sent to your email.",
    "Yes, we ship to selected countries. Shipping charges vary by destination.",
    "You can contact us via email, live chat, or our customer support phone number.",
    "Yes, cash on delivery is available for eligible locations.",
    "Click 'Sign Up' on the homepage and fill in your details to create a new account.",
    "Yes, we use industry-standard encryption to securely store your card details.",
    "You can change your delivery address before the order is shipped, from the 'My Orders' section.",
    "No, we do not charge any extra fees for online payments.",
    "Please check your card details and try again, or contact your bank if the issue persists.",
    "Enter your coupon code at checkout in the 'Apply Coupon' field before placing your order.",
    "Yes, we support payments through major mobile wallets.",
    "Go to Account Settings and select 'Delete Account' to permanently remove your account."
]
```

```python
def clean_question(raw_text):
    text = raw_text.lower()

    tokens = word_tokenize(text)

    stop_words = set(stopwords.words('english'))

    cleaned_text = [
        word for word in tokens
        if word not in stop_words and word not in string.punctuation
    ]

    return " ".join(cleaned_text)
```

```python
cleaned_faqs = [clean_question(faq) for faq in FAQs]

vectorizer = TfidfVectorizer()
faq_matrix = vectorizer.fit_transform(cleaned_faqs)
```

```python
print("Hello! Ask me anything about our online store.")
print("(Type 'exit' to end the chat)\n")

while True:
    user_question = input("You: ")

    # Handle empty input
    if user_question.strip() == "":
        print("Bot: Please type a question, or type 'exit' to leave.\n")
        continue

    # Exit condition
    if user_question.strip().lower() == "exit":
        print("\nBot: Thank you for using the chatbot!")
        break

    cleaned_question = clean_question(user_question)

    # Handle input that becomes empty after cleaning (e.g. only stopwords/punctuation)
    if cleaned_question.strip() == "":
        print("Bot: Sorry, I didn't understand that. Could you rephrase?\n")
        continue

    user_question_vector = vectorizer.transform([cleaned_question])
    similarity_scores = cosine_similarity(user_question_vector, faq_matrix)

    best_match_index = np.argmax(similarity_scores)
    best_score = similarity_scores[0][best_match_index]

    if best_score < 0.50:
        print("Bot: Sorry, your question doesn't seem related to our store. Could you rephrase it?\n")
    else:
        print(f"Bot: {Answers[best_match_index]}\n")
```
